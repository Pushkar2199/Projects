```python
#handling missing values
df = pd.read_csv('orders.csv', na_values = ['Not Available','unknown'] )

df['Ship Mode'].unique()
```

```python
#rename columns names ..make them lower case and replace space with underscore

df.columns = df.columns.str.replace(' ','_').str.lower()
```

```python
df.columns
```

```python
df.head(5)
```

```python
#derive new columns discount

df['discount'] =df['list_price']*df['discount_percent']/100

df[['list_price','discount_percent','discount']].head(5)
```

```python
#derive new columns sale price

df['sale_price']= df['list_price']-df['discount']
df[['list_price',  'discount' ,'sale_price']].head()
```

```python
#derive new columns profit

df['profit'] = df['sale_price'] - df['cost_price']

df[['cost_price','sale_price','profit']].head()
```

```python
#convert order date from object data type to datetime
df['order_date']=pd.to_datetime(df['order_date'],format="%Y-%m-%d")
```
